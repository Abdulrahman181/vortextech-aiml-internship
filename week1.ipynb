```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os
from pathlib import Path

configured_csv = os.environ.get('TITANIC_TRAIN_CSV')
candidate_paths = (
    [Path(configured_csv)]
    if configured_csv
    else [
        Path('/kaggle/input/competitions/titanic/train.csv'),
        Path('data/train.csv'),
        Path('train.csv'),
    ]
)
train_csv = next((path for path in candidate_paths if path.is_file()), None)
if train_csv is None:
    searched_paths = '\n'.join(f'  - {path}' for path in candidate_paths)
    raise FileNotFoundError(
        'Titanic training CSV not found. Set TITANIC_TRAIN_CSV to its path or '
        'place it at data/train.csv or train.csv. Searched:\n'
        f'{searched_paths}'
    )

df = pd.read_csv(train_csv)

df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.drop(columns=['Cabin'], inplace=True, errors='ignore')
```

```python
df.drop_duplicates(inplace=True)
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df['Age'], bins=30, edgecolor='black')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.show()
```
